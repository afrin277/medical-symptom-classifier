## Importing libraries

```python
import pandas as pd
import numpy as np
import os
import joblib

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split

print("Libraries imported successfully!")
```

## Loading the encoded dataset

```python
df = pd.read_csv("../data/processed/feature_dataset.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

### Separating X and y

```python
X = df.drop("Disease", axis=1)

y = df["Disease"]

print("Input features shape:", X.shape)
print("Target shape:", y.shape)

print("\nInput features:")
print(X.columns.tolist()[:10])

print("\nTarget samples:")
print(y.head())
```

```python
label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Label encoding completed!")

print("Number of disease classes:",
      len(label_encoder.classes_))

print("\nDisease label mapping:")

for index, disease in enumerate(label_encoder.classes_):
    print(index, ":", disease)
```

### Checking class distribution before splitting

```python
class_counts = y.value_counts()

print("Minimum records per disease:",
      class_counts.min())

print("\nDiseases with fewer than 2 records:")

print(class_counts[class_counts < 2])

print("\nTotal disease classes:", len(class_counts))
```

### Splitting the dataset

```python
if class_counts.min() < 2:
    print("Some disease classes have fewer than 2 records.")
    print("We need to review those classes before stratified splitting.")

else:
    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y_encoded,
        test_size=0.20,
        random_state=42,
        stratify=y_encoded
    )

    print("Dataset splitting completed!")

    print("Training features:", X_train.shape)
    print("Testing features:", X_test.shape)

    print("Training targets:", y_train.shape)
    print("Testing targets:", y_test.shape)
```

### Feature scaling

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)

X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")

print("Scaled training shape:", X_train_scaled.shape)
print("Scaled testing shape:", X_test_scaled.shape)
```

### Verifying the prepared data

```python
print("TRAINING DATA SUMMARY")
print("-" * 40)

print("X_train:", X_train_scaled.shape)
print("y_train:", y_train.shape)

print("\nTESTING DATA SUMMARY")
print("-" * 40)

print("X_test:", X_test_scaled.shape)
print("y_test:", y_test.shape)

print("\nNumber of training classes:",
      len(np.unique(y_train)))

print("Number of testing classes:",
      len(np.unique(y_test)))
```

### Saving prepared data and encoders

```python
os.makedirs("../data/processed", exist_ok=True)
os.makedirs("../models", exist_ok=True)

np.save("../data/processed/X_train.npy", X_train_scaled)
np.save("../data/processed/X_test.npy", X_test_scaled)

np.save("../data/processed/y_train.npy", y_train)
np.save("../data/processed/y_test.npy", y_test)

joblib.dump(label_encoder, "../models/label_encoder.pkl")
joblib.dump(scaler, "../models/feature_scaler.pkl")

joblib.dump(X.columns.tolist(), "../models/symptom_columns.pkl")

print("Prepared datasets and encoders saved successfully!")
```
