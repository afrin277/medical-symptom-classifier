### Identifying the target and feature columns

```python
target_column = "Disease"

symptom_columns = [
    col for col in df.columns
    if col.startswith("Symptom_")
]

print("Target column:", target_column)
print("Number of symptom columns:", len(symptom_columns))
print("Symptom columns:", symptom_columns)
```

### Checking the missing values

```python
missing = df.isnull().sum()

missing_df = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (missing / len(df)) * 100
})

missing_df
```

### Visualising missing values

```python
plt.figure(figsize=(12, 5))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Missing Values in Original Dataset")
plt.xlabel("Columns")
plt.tight_layout()
plt.show()
```

### Checking the duplicate records

```python
print("Exact duplicate rows:", df.duplicated().sum())
```

```python
symptom_pattern_duplicates = df.duplicated(
    subset=symptom_columns,
    keep=False
)

print(
    "Rows with repeated symptom combinations:",
    symptom_pattern_duplicates.sum()
)
```

### Understanding the disease distribution

```python
disease_counts = df[target_column].value_counts()

print("Number of disease classes:", df[target_column].nunique())

disease_counts
```

### Visualisation

```python
plt.figure(figsize=(15, 7))

disease_counts.plot(kind="bar")

plt.title("Distribution of Diseases")
plt.xlabel("Disease")
plt.ylabel("Number of Records")
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

### Check disease names across files

```python
main_diseases = set(df["Disease"].str.strip())

description_diseases = set(
    description_df["Disease"].str.strip()
)

precaution_diseases = set(
    precaution_df["Disease"].str.strip()
)

print("Diseases in main dataset:", len(main_diseases))
print("Diseases in description file:", len(description_diseases))
print("Diseases in precaution file:", len(precaution_diseases))

print("\nMissing descriptions:")
print(main_diseases - description_diseases)

print("\nMissing precautions:")
print(main_diseases - precaution_diseases)
```
